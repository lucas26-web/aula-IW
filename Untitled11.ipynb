```python
#QUESTÃO 02 – Soma dos Números
#---------------------------------------------------------

#Solicite ao usuário um número inteiro positivo.

#Utilizando um laço de repetição, calcule e exiba a soma de
#todos os números de 1 até o número informado.

#Exemplo:

#Digite um número: 5
#Resultado: 15
#Pois:
#1 + 2 + 3 + 4 + 5 = 15

#Entrada

numero = int(input("Digite um numero: "))
soma = 0
for i in range (1, numero +1):
  soma += i
  print(soma)
  if i == numero:
    print(soma)

```

```python
#QUESTÃO 03 – Tabuada
#---------------------------------------------------------

#Solicite ao usuário um número inteiro e exiba sua tabuada de 1 até 10.

#Exemplo:

#Digite um número: 7

#7 x 1 = 7
#7 x 2 = 14
#7 x 3 = 21
#...
#7 x 10 = 70

#Entrada


num = int(input("Digite um número: "))
print(f"A tabuada do {num} é:")

for i in range(1, 11):
    resultado = num * i

    print(f"{num} x {i} = {resultado}")

print("=========================================================================")
```

```python
#QUESTÃO 04 – Fatorial
#---------------------------------------------------------

#Solicite ao usuário um número inteiro positivo e calcule seu fatorial utilizando um laço de repetição.

#Exemplo:

#Digite um número: 5

#Fatorial = 120

#Pois:

#5! = 5 × 4 × 3 × 2 × 1

#Observação:
#O programa deve validar se o número informado é positivo.

#entrada

while True:
    numero = int(input("Digite um número inteiro positivo: "))
    if numero >= 0:
        break
    print("Número inválido, digite um número maior ou igual a zero.")

fatorial = 1

for i in range(1, numero + 1):
    fatorial *= i

print(f"Fatorial = {fatorial}")
```

```python
# QUESTÃO 05 – Caixa Eletrônico (5 pontos)
# ---------------------------------------------------------

# Solicita ao usuário o valor do saque

valor_saque = int(input("Digite o valor do saque: "))
valor_restante = valor_saque

print("\nSaída esperada:")

# Processa notas de R$100
if valor_restante >= 100:
    quantidade = 0
    while valor_restante >= 100:
        quantidade += 1
        valor_restante -= 100
    print(f"{quantidade} nota(s) de R$100")

# Processa notas de R$50
if valor_restante >= 50:
    quantidade = 0
    while valor_restante >= 50:
        quantidade += 1
        valor_restante -= 50
    print(f"{quantidade} nota(s) de R$50")

# Processa notas de R$20
if valor_restante >= 20:
    quantidade = 0
    while valor_restante >= 20:
        quantidade += 1
        valor_restante -= 20
    print(f"{quantidade} nota(s) de R$20")

# Processa notas de R$10
if valor_restante >= 10:
    quantidade = 0
    while valor_restante >= 10:
        quantidade += 1
        valor_restante -= 10
    print(f"{quantidade} nota(s) de R$10")

# Processa notas de R$5
if valor_restante >= 5:
    quantidade = 0
    while valor_restante >= 5:
        quantidade += 1
        valor_restante -= 5
    print(f"{quantidade} nota(s) de R$5")

# Processa notas de R$2
if valor_restante >= 2:
    quantidade = 0
    while valor_restante >= 2:
        quantidade += 1
        valor_restante -= 2
    print(f"{quantidade} nota(s) de R$2")

# Processa notas de R$1
if valor_restante >= 1:
    quantidade = 0
    while valor_restante >= 1:
        quantidade += 1
        valor_restante -= 1
    print(f"{quantidade} nota(s) de R$1")
```
